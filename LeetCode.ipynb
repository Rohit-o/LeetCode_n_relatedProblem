**Palindromic Substrings Counting**

Let say you have string s1 = "babad", then how many palindromic is possible here

```python
def count_PalindromicSubstrings (s : str) -> int:
  count = 0
  n = len(s)
  def ExpandAndCount (left : int, right : int) -> int:
    count_at_centre = 0
    L = left
    R = right
    while L >= 0 and R < n and s[L] == s[R]:
      count_at_centre += 1
      L -= 1
      R += 1
    return count_at_centre
  for i in range(n):
      count += ExpandAndCount(i, i)
      count += ExpandAndCount(i, i + 1)
  return count
  count
s1 = "racecar"
# Palindromes: "b", "a", "b", "a", "d", "bab", "aba"
print(f"String: '{s1}'")
print(f"Total palindromic substrings: {count_PalindromicSubstrings(s1)}")
# Output for "babad" is 7
```

This same thing can be done using the class.

```python
class PalindromicSubstringsCounter:
    """
    A class to count all palindromic substrings within a given string.
    """
    def __init__(self, s: str):
        self.s = s
        self.n = len(s)
        self.count = 0

    def _expand_and_count(self, left: int, right: int) -> None:
        """
        Expands from a center (or centers) and counts valid palindromes.
        Updates the total count stored in the instance attribute.
        """
        L = left
        R = right
        while L >= 0 and R < self.n and self.s[L] == self.s[R]:
            self.count += 1
            L -= 1
            R += 1

    def count_palindromes(self) -> int:
        """
        Iterates through the string to find all palindromes by considering
        each character as a center for odd-length and even-length palindromes.
        """
        # Iterate through each potential center point
        for i in range(self.n):
            # Case 1: Odd length palindromes (center is a single character s[i])
            self._expand_and_count(i, i)
            # Case 2: Even length palindromes (center is between s[i] and s[i+1])
            self._expand_and_count(i, i + 1)
        return self.count

# --- Example Usage ---
s1 = "racecar"
counter = PalindromicSubstringsCounter(s1)
total_palindromes = counter.count_palindromes()

print(f"String: '{s1}'")
print(f"Total palindromic substrings: {total_palindromes}")

s2 = "babad"
counter2 = PalindromicSubstringsCounter(s2)
print(f"\nString: '{s2}'")
print(f"Total palindromic substrings: {counter2.count_palindromes()}") # Output for "babad" is 7
```

Given a string s, return the longest palindromic substring in s.

Example 1:
Input: s = "babad"
Output: "bab"
Explanation: "aba" is also a valid answer.

Example 2:
Input: s = "cbbd"
Output: "bb"

Constraints:
1 <= s.length <= 1000
s consist of only digits and English letters.

```python
class Solution:
    def longestPalindrome(self, s: str) -> str:
      self.start = 0
      self.max_length = 0
      N = len(s)

      def expand_around_center(left: int, right: int):
        while left >= 0 and right < N and s[left] == s[right]:
          current_length = right - left + 1
          if current_length > self.max_length:
            self.max_length = current_length
            self.start = left
          left = left - 1
          right = right + 1
      for i in range (N):
        expand_around_center(i,i)
        expand_around_center(i,i+1)
      return s[self.start : self.start + self.max_length]

sol = Solution()
s1 = 'bababd'
print(f"Input: {s1}, Longest Palindrome: '{sol.longestPalindrome(s1)}'")

sol2 = Solution() # Use a new instance for a new input to reset internal state
s2 = "cbbd"
print(f"Input: {s2}, Longest Palindrome: '{sol2.longestPalindrome(s2)}'")
```

#Problem 241
You are given a string s. You can convert s to a palindrome by **adding** characters in front of it.

Return the shortest palindrome you can find by performing this transformation.

Example 1:

Input: s = "aacecaaa"
Output: "aaacecaaa"
Example 2:

Input: s = "abcd"
Output: "dcbabcd"

Constraints:

0 <= s.length <= 5 * 104
s consists of lowercase English letters only.

```python
class Solution:
    def shortestPalindrome(self, s: str) -> str:
        n = len(s)
        i = 0      # Pointer from the start
        j = n - 1  # Pointer from the end

        # Keep J moving backward until we find where the longest palindromic
        # prefix (starting at index 0) ends.
        while j >= 0:
            if s[i] == s[j]:
                i += 1  # We found another matching character for our prefix
            j -= 1

        # At this point, 'i' is the length of the longest palindromic prefix
        # that started at index 0.

        # If 'i' is equal to the length of the string 'n',
        # the entire string 's' is already a palindrome.
        if i == n:
            return s

        # The remaining non-palindromic suffix starts from index 'i' to the end.
        suffix = s[i:]

        # We reverse this suffix and prepend it to the original string 's'.
        reversed_suffix = suffix[::-1]

        return reversed_suffix + s

# --- Example Usage ---
sol = Solution()

s1 = "aababa"
# Longest prefix is "aacecaa" (length i=7)
# Suffix is "a" (s[7:])
# Reversed suffix is "a"
# Result: "a" + "aacecaaa" = "aaacecaaa"
print(f"Input: {s1}, Output: {sol.shortestPalindrome(s1)}")

s2 = "abcd"
# Longest prefix is "" (length i=0, mismatch immediately)
# Suffix is "abcd" (s[0:])
# Reversed suffix is "dcba"
# Result: "dcba" + "abcd" = "dcbabcd"
print(f"Input: {s2}, Output: {sol.shortestPalindrome(s2)}")
```

```python
class Solution:
    def shortestPalindrome(self, s: str) -> str:
        rs = s[::-1]
        t = s + "#" + rs
        # Build LPS array (KMP prefix function)
        lps = [0] * len(t)
        j = 0
        for i in range(1, len(t)):
            while j > 0 and t[i] != t[j]:
                j = lps[j - 1]
            if t[i] == t[j]:
                j += 1
                lps[i] = j
        pal_pref_len = lps[-1]          # longest palindromic prefix length in s
        add = s[pal_pref_len:][::-1]    # reverse of the remaining suffix
        return add + s
sol = Solution()
s1 = "aababa"
print(f"Input: {s1}, Output: {sol.shortestPalindrome(s1)}")
```

```python
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        # Dictionary to store value: index pairs
        prevMap = {}

        for i, n in enumerate(nums):
            diff = target - n
            # If the complement (diff) is already in our map, we found the pair
            if diff in prevMap:
                return [prevMap[diff], i]
            # Otherwize, add current number and its index to the map
            prevMap[n] = i

sol = Solution()
result = sol.twoSum([1, 3, 4, 11], 7)
print(f"{result}")
```

```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def addTwoNumbers(self, l1: ListNode, l2: ListNode) -> ListNode:
        dummy = ListNode(0)
        tail = dummy
        carry = 0

        while l1 is not None or l2 is not None or carry != 0:
            x = l1.val if l1 is not None else 0
            y = l2.val if l2 is not None else 0

            total = x + y + carry
            carry = total // 10
            digit = total % 10

            tail.next = ListNode(digit)
            tail = tail.next

            if l1 is not None:
                l1 = l1.next
            if l2 is not None:
                l2 = l2.next

        return dummy.next

def create_linked_list(arr):
    """Converts a Python list into a ListNode-based linked list."""
    dummy = ListNode(0)
    current = dummy
    for val in arr:
        current.next = ListNode(val)
        current = current.next
    return dummy.next

def print_linked_list(node):
    """Prints the linked list in a readable format (e.g., 7 -> 0 -> 8)."""
    values = []
    while node:
        values.append(str(node.val))
        node = node.next
    print(" -> ".join(values) if values else "Empty List")


# 1. Initialize the solution class
sol = Solution()

# 2. Create test inputs
# Example: 342 + 465 = 807
l1 = create_linked_list([2, 4, 3])
l2 = create_linked_list([5, 6, 4])

# 3. Call the method
result = sol.addTwoNumbers(l1, l2)

# 4. See if it works!
print("Result:")
print_linked_list(result)  # Expected Output: 7 -> 0 -> 8
```

```python
from typing import Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
        dummy = ListNode(0)
        tail = dummy
        carry = 0
        while l1 or l2 or carry:
            if l1:
                carry += l1.val
                l1 = l1.next
            if l2:
                carry += l2.val
                l2 = l2.next
            tail.next = ListNode(carry % 10)
            tail = tail.next
            carry //= 10
        return dummy.next

def create_linked_list(arr):
    """Converts a Python list into a ListNode-based linked list."""
    dummy = ListNode(0)
    current = dummy
    for val in arr:
        current.next = ListNode(val)
        current = current.next
    return dummy.next

def print_linked_list(node):
    """Prints the linked list in a readable format (e.g., 7 -> 0 -> 8)."""
    values = []
    while node:
        values.append(str(node.val))
        node = node.next
    print(" -> ".join(values) if values else "Empty List")


# 1. Initialize the solution class
sol = Solution()

# 2. Create test inputs
# Example: 342 + 465 = 807
l1 = create_linked_list([2, 4, 3])
l2 = create_linked_list([5, 6, 4])

# 3. Call the method
result = sol.addTwoNumbers(l1, l2)

# 4. See if it works!
print("Result:")
print_linked_list(result)  # Expected Output: 7 -> 0 -> 8
```

You are given two strings word1 and word2. Merge the strings by adding letters in alternating order, starting with word1. If a string is longer than the other, append the additional letters onto the end of the merged string.

Return the merged string.



Example 1:

Input: word1 = "abc", word2 = "pqr"
Output: "apbqcr"
Explanation: The merged string will be merged as so:
word1:  a   b   c
word2:    p   q   r
merged: a p b q c r
Example 2:

Input: word1 = "ab", word2 = "pqrs"
Output: "apbqrs"
Explanation: Notice that as word2 is longer, "rs" is appended to the end.
word1:  a   b
word2:    p   q   r   s
merged: a p b q   r   s
Example 3:

Input: word1 = "abcd", word2 = "pq"
Output: "apbqcd"
Explanation: Notice that as word1 is longer, "cd" is appended to the end.
word1:  a   b   c   d
word2:    p   q
merged: a p b q c   d


Constraints:

1 <= word1.length, word2.length <= 100
word1 and word2 consist of lowercase English letters.

```python
class Solution:
    def mergeAlternately(self, word1: str, word2: str) -> str:
        merged_chars = []
        min_len = min(len(word1), len(word2))

        for i in range (min_len):
            merged_chars.append(word1[i])
            merged_chars.append(word2[i])

        if len(word1) > len(word2):
            merged_chars.extend(word1)
        else:
            merged_chars.extend(word2)

        return "".join(merged_chars)

sol = Solution()
result = sol.mergeAlternately("abcd", "pqr")
print(result) # Output: 'apbqrd'
```
