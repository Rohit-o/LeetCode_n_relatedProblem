**Palindromic Substrings Counting**

Let say you have string s1 = "babad", then how many palindromic is possible here

```python
def count_PalindromicSubstrings (s : str) -> int:
  count = 0
  n = len(s)
  def ExpandAndCount (left : int, right : int) -> int:
    count_at_centre = 0
    L = left
    R = right
    while L >= 0 and R < n and s[L] == s[R]:
      count_at_centre += 1
      L -= 1
      R += 1
    return count_at_centre
  for i in range(n):
      count += ExpandAndCount(i, i)
      count += ExpandAndCount(i, i + 1)
  return count
  count
s1 = "racecar"
# Palindromes: "b", "a", "b", "a", "d", "bab", "aba"
print(f"String: '{s1}'")
print(f"Total palindromic substrings: {count_PalindromicSubstrings(s1)}")
# Output for "babad" is 7
```

This same thing can be done using the class.

```python
class PalindromicSubstringsCounter:
    """
    A class to count all palindromic substrings within a given string.
    """
    def __init__(self, s: str):
        self.s = s
        self.n = len(s)
        self.count = 0

    def _expand_and_count(self, left: int, right: int) -> None:
        """
        Expands from a center (or centers) and counts valid palindromes.
        Updates the total count stored in the instance attribute.
        """
        L = left
        R = right
        while L >= 0 and R < self.n and self.s[L] == self.s[R]:
            self.count += 1
            L -= 1
            R += 1

    def count_palindromes(self) -> int:
        """
        Iterates through the string to find all palindromes by considering
        each character as a center for odd-length and even-length palindromes.
        """
        # Iterate through each potential center point
        for i in range(self.n):
            # Case 1: Odd length palindromes (center is a single character s[i])
            self._expand_and_count(i, i)
            # Case 2: Even length palindromes (center is between s[i] and s[i+1])
            self._expand_and_count(i, i + 1)
        return self.count

# --- Example Usage ---
s1 = "racecar"
counter = PalindromicSubstringsCounter(s1)
total_palindromes = counter.count_palindromes()

print(f"String: '{s1}'")
print(f"Total palindromic substrings: {total_palindromes}")

s2 = "babad"
counter2 = PalindromicSubstringsCounter(s2)
print(f"\nString: '{s2}'")
print(f"Total palindromic substrings: {counter2.count_palindromes()}") # Output for "babad" is 7
```

Given a string s, return the longest palindromic substring in s.

Example 1:
Input: s = "babad"
Output: "bab"
Explanation: "aba" is also a valid answer.

Example 2:
Input: s = "cbbd"
Output: "bb"

Constraints:
1 <= s.length <= 1000
s consist of only digits and English letters.

```python
class Solution:
    def longestPalindrome(self, s: str) -> str:
      self.start = 0
      self.max_length = 0
      N = len(s)

      def expand_around_center(left: int, right: int):
        while left >= 0 and right < N and s[left] == s[right]:
          current_length = right - left + 1
          if current_length > self.max_length:
            self.max_length = current_length
            self.start = left
          left = left - 1
          right = right + 1
      for i in range (N):
        expand_around_center(i,i)
        expand_around_center(i,i+1)
      return s[self.start : self.start + self.max_length]

sol = Solution()
s1 = 'bababd'
print(f"Input: {s1}, Longest Palindrome: '{sol.longestPalindrome(s1)}'")

sol2 = Solution() # Use a new instance for a new input to reset internal state
s2 = "cbbd"
print(f"Input: {s2}, Longest Palindrome: '{sol2.longestPalindrome(s2)}'")
```

#Problem 241
You are given a string s. You can convert s to a palindrome by **adding** characters in front of it.

Return the shortest palindrome you can find by performing this transformation.

Example 1:

Input: s = "aacecaaa"
Output: "aaacecaaa"
Example 2:

Input: s = "abcd"
Output: "dcbabcd"

Constraints:

0 <= s.length <= 5 * 104
s consists of lowercase English letters only.

```python
class Solution:
    def shortestPalindrome(self, s: str) -> str:
        n = len(s)
        i = 0      # Pointer from the start
        j = n - 1  # Pointer from the end

        # Keep J moving backward until we find where the longest palindromic
        # prefix (starting at index 0) ends.
        while j >= 0:
            if s[i] == s[j]:
                i += 1  # We found another matching character for our prefix
            j -= 1

        # At this point, 'i' is the length of the longest palindromic prefix
        # that started at index 0.

        # If 'i' is equal to the length of the string 'n',
        # the entire string 's' is already a palindrome.
        if i == n:
            return s

        # The remaining non-palindromic suffix starts from index 'i' to the end.
        suffix = s[i:]

        # We reverse this suffix and prepend it to the original string 's'.
        reversed_suffix = suffix[::-1]

        return reversed_suffix + s

# --- Example Usage ---
sol = Solution()

s1 = "aababa"
# Longest prefix is "aacecaa" (length i=7)
# Suffix is "a" (s[7:])
# Reversed suffix is "a"
# Result: "a" + "aacecaaa" = "aaacecaaa"
print(f"Input: {s1}, Output: {sol.shortestPalindrome(s1)}")

s2 = "abcd"
# Longest prefix is "" (length i=0, mismatch immediately)
# Suffix is "abcd" (s[0:])
# Reversed suffix is "dcba"
# Result: "dcba" + "abcd" = "dcbabcd"
print(f"Input: {s2}, Output: {sol.shortestPalindrome(s2)}")
```

```python
class Solution:
    def shortestPalindrome(self, s: str) -> str:
        rs = s[::-1]
        t = s + "#" + rs
        # Build LPS array (KMP prefix function)
        lps = [0] * len(t)
        j = 0
        for i in range(1, len(t)):
            while j > 0 and t[i] != t[j]:
                j = lps[j - 1]
            if t[i] == t[j]:
                j += 1
                lps[i] = j
        pal_pref_len = lps[-1]          # longest palindromic prefix length in s
        add = s[pal_pref_len:][::-1]    # reverse of the remaining suffix
        return add + s
sol = Solution()
s1 = "aababa"
print(f"Input: {s1}, Output: {sol.shortestPalindrome(s1)}")
```

```python
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        # Dictionary to store value: index pairs
        prevMap = {}

        for i, n in enumerate(nums):
            diff = target - n
            # If the complement (diff) is already in our map, we found the pair
            if diff in prevMap:
                return [prevMap[diff], i]
            # Otherwize, add current number and its index to the map
            prevMap[n] = i

sol = Solution()
result = sol.twoSum([1, 3, 4, 11], 7)
print(f"{result}")
```

Constraints:

word1 and word2 consist of lowercase English letters.

```python
class Solution:
    def mergeAlternately(self, word1: str, word2: str) -> str:
        merged_chars = []
        min_len = min(len(word1), len(word2))

        for i in range (min_len):
            merged_chars.append(word1[i])
            merged_chars.append(word2[i])

        if len(word1) > len(word2):
            merged_chars.extend(word1)
        else:
            merged_chars.extend(word2)

        return "".join(merged_chars)

sol = Solution()
result = sol.mergeAlternately("abcd", "pqr")
print(result) # Output: 'apbqrd'
```

##3651. Minimum Cost Path with Teleportations

You are given a m x n 2D integer array grid and an integer k. You start at the top-left cell (0, 0) and your goal is to reach the bottom‐right cell (m - 1, n - 1).

There are two types of moves available:

Normal move: You can move right or down from your current cell (i, j), i.e. you can move to (i, j + 1) (right) or (i + 1, j) (down). The cost is the value of the destination cell.

Teleportation: You can teleport from any cell (i, j), to any cell (x, y) such that grid[x][y] <= grid[i][j]; the cost of this move is 0. You may teleport at most k times.

Return the minimum total cost to reach cell (m - 1, n - 1) from (0, 0).



Example 1:

Input: grid = [[1,3,3],[2,5,4],[4,3,5]], k = 2

Output: 7

Explanation:

Initially we are at (0, 0) and cost is 0.

Current Position	Move	New Position	Total Cost
(0, 0)	Move Down	(1, 0)	0 + 2 = 2
(1, 0)	Move Right	(1, 1)	2 + 5 = 7
(1, 1)	Teleport to (2, 2)	(2, 2)	7 + 0 = 7
The minimum cost to reach bottom-right cell is 7.

Example 2:

Input: grid = [[1,2],[2,3],[3,4]], k = 1

Output: 9

Explanation:

Initially we are at (0, 0) and cost is 0.

Current Position	Move	New Position	Total Cost
(0, 0)	Move Down	(1, 0)	0 + 2 = 2
(1, 0)	Move Right	(1, 1)	2 + 3 = 5
(1, 1)	Move Down	(2, 1)	5 + 4 = 9
The minimum cost to reach bottom-right cell is 9.



Constraints:

2 <= m, n <= 80
m == grid.length
n == grid[i].length
0 <= grid[i][j] <= 104
0 <= k <= 10

```python
# Brute Force Method
def minPathCost(grid, k):
    m, n = len(grid), len(grid[0])
    # dp[p][i][j] = min cost to reach (i, j) using p teleports
    dp = [[[float('inf')] * n for _ in range(m)] for _ in range(k + 1)]

    # Starting point: cost is 0 as per explanation
    dp[0][0][0] = 0

    for p in range(k + 1):
        # 1. Handle Normal Moves (Right and Down)
        for i in range(m):
            for j in range(n):
                if i == 0 and j == 0: continue

                # Move from Top
                if i > 0:
                    dp[p][i][j] = min(dp[p][i][j], dp[p][i-1][j] + grid[i][j])
                # Move from Left
                if j > 0:
                    dp[p][i][j] = min(dp[p][i][j], dp[p][i][j-1] + grid[i][j])

        # 2. Handle Teleportation (to be used for the NEXT teleport level p+1)
        if p < k:
            # We can teleport from ANY cell (i, j) reached with p teleports
            # to ANY cell (x, y) if grid[x][y] <= grid[i][j] at 0 cost.

            # Find the best (minimum) cost available for each grid value
            # This is an optimization to avoid O(N^4) comparison
            best_cost_for_val = float('inf')

            # Sort all cells by their grid value descending to check teleport eligibility
            all_cells = []
            for r in range(m):
                for c in range(n):
                    if dp[p][r][c] != float('inf'):
                        all_cells.append((grid[r][c], dp[p][r][c]))

            # For each potential destination (x, y)
            for x in range(m):
                for y in range(n):
                    dest_val = grid[x][y]
                    # Check if any source (i, j) exists where grid[i][j] >= dest_val
                    for src_val, src_cost in all_cells:
                        if src_val >= dest_val:
                            dp[p+1][x][y] = min(dp[p+1][x][y], src_cost)

    # The answer is the minimum cost at the destination across all possible teleport counts
    ans = min(dp[p][m-1][n-1] for p in range(k + 1))
    return ans

# Test Case
grid = [[1, 3, 3], [2, 5, 4], [4, 3, 5]]
k = 2
print(f"Minimum Cost: {minPathCost(grid, k)}") # Expected Output: 7

```

```python
# ---------- SOLUTION CODE ----------

import heapq
from bisect import bisect_right

class DSUNext:
    def __init__(self, n):
        self.parent = list(range(n + 1))  # sentinel at n

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]

    def remove(self, x):
        self.parent[x] = self.find(x + 1)

class Solution:
    def minCost(self, grid, k):
        m, n = len(grid), len(grid[0])
        N = m * n

        def idx(i, j):
            return i * n + j

        vals = [0] * N
        for i in range(m):
            for j in range(n):
                vals[idx(i, j)] = grid[i][j]

        order = list(range(N))
        order.sort(key=lambda x: vals[x])
        sorted_vals = [vals[i] for i in order]

        INF = 10**15
        dist = [[INF] * N for _ in range(k + 1)]
        dist[0][0] = 0

        dsu_layer = [None] * (k + 1)
        for t in range(1, k + 1):
            dsu_layer[t] = DSUNext(N)

        pq = [(0, 0, 0)]  # (cost, cell_id, teleports_used)

        while pq:
            cost, u, t = heapq.heappop(pq)
            if cost != dist[t][u]:
                continue

            i, j = divmod(u, n)

            # Normal moves
            if j + 1 < n:
                v = u + 1
                nc = cost + vals[v]
                if nc < dist[t][v]:
                    dist[t][v] = nc
                    heapq.heappush(pq, (nc, v, t))

            if i + 1 < m:
                v = u + n
                nc = cost + vals[v]
                if nc < dist[t][v]:
                    dist[t][v] = nc
                    heapq.heappush(pq, (nc, v, t))

            # Teleport
            if t < k:
                r = bisect_right(sorted_vals, vals[u]) - 1
                if r >= 0:
                    dsu = dsu_layer[t + 1]
                    p = dsu.find(0)
                    while p <= r:
                        v = order[p]
                        if cost < dist[t + 1][v]:
                            dist[t + 1][v] = cost
                            heapq.heappush(pq, (cost, v, t + 1))
                        dsu.remove(p)
                        p = dsu.find(p)

        target = N - 1
        return min(dist[t][target] for t in range(k + 1))


# ---------- HOW TO CALL IT ----------

grid = [[1, 3, 3],
        [2, 5, 4],
        [4, 3, 5]]
k = 2

sol = Solution()
answer = sol.minCost(grid, k)

print("Minimum Cost:", answer)
```

## 1979. Find Greatest Common Divisor of Array

Given an integer array nums, return the greatest common divisor of the smallest number and largest number in nums.

The greatest common divisor of two numbers is the largest positive integer that evenly divides both numbers.



Example 1:

Input: nums = [2,5,6,9,10]
Output: 2
Explanation:
The smallest number in nums is 2.
The largest number in nums is 10.
The greatest common divisor of 2 and 10 is 2.
Example 2:

Input: nums = [7,5,6,8,3]
Output: 1
Explanation:
The smallest number in nums is 3.
The largest number in nums is 8.
The greatest common divisor of 3 and 8 is 1.
Example 3:

Input: nums = [3,3]
Output: 3
Explanation:
The smallest number in nums is 3.
The largest number in nums is 3.
The greatest common divisor of 3 and 3 is 3.


Constraints:

2 <= nums.length <= 1000
1 <= nums[i] <= 1000

```python
from typing import List

class Solution:
    def findGCD(self, nums: List[int]) -> int:
        b = min(nums)
        a = max(nums)
        while b != 0:
            a, b = b, a % b  # Fixed the swap here
        return a

# --- How to use it ---

# 1. Create an instance of the class
solver = Solution()

# 2. Define your input
my_nums = [2, 5, 6, 9, 10]

# 3. Call the function and print result
result = solver.findGCD(my_nums)
print(f"The GCD is: {result}")
```
